## (c)

```python
rng = np.random.default_rng()
```

```python
def rcnorm(n, mean, sd, c):
    out = np.zeros(n)
    for i in range(n):
        while True:
            out[i] = rng.normal(mean, sd, 1).item()
            if out[i]>c:
                break
    return out
```

```python
rcnorm(10, 1, 2, 1)
```

## (e)

```python
def integrate_trapezoid(f, lower, upper, n):
    delta = (upper-lower) / (n-1)
    x = np.linspace(lower, upper, n)
    return delta * sum([(f(x[i]) + f(x[i+1]))/2 for i in range(n-1)])
```

```python
def integrate_mc(f, lower, upper, n):
    return np.mean(list(map(f, rng.uniform(lower, upper, n))))
```

```python
from scipy.integrate import quad
def intergate_scipy(f, lower, upper):
    return quad(f, lower, upper)[0]
```

```python
print(integrate_trapezoid(np.sin, 0, 1, 100))
print(integrate_mc(np.sin, 0, 1, 100))
print(intergate_scipy(np.sin, 0, 1))
```
